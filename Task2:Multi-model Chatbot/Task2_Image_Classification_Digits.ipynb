# Task 2 — Multimodal Demo (Digits Image Classification)

This notebook demonstrates a simple image-classification pipeline using sklearn's digits dataset. It's a reproducible example for visual outputs and model comparisons.

```
# Imports and data
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay

digits = load_digits()
X = digits.images
y = digits.target

# Flatten images for classic ML models
n_samples = len(X)
X_flat = X.reshape((n_samples, -1))

X_train, X_test, y_train, y_test = train_test_split(X_flat, y, test_size=0.2, random_state=42, stratify=y)

# Train simple models
lr = LogisticRegression(max_iter=1000, multi_class='auto', solver='lbfgs')
rf = RandomForestClassifier(n_estimators=200, random_state=42)
lr.fit(X_train, y_train); rf.fit(X_train, y_train)

# Evaluate
y_pred_lr = lr.predict(X_test); y_pred_rf = rf.predict(X_test)
print('Logistic Regression:\n', classification_report(y_test, y_pred_lr))
print('Random Forest:\n', classification_report(y_test, y_pred_rf))

# Save confusion matrix plots
cm = confusion_matrix(y_test, y_pred_rf)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=digits.target_names)
fig = disp.plot().figure_
fig.suptitle('Digits - Random Forest Confusion Matrix')
fig.savefig('task2_confusion_rf.png')
```
