# Task 5 — Domain Expert: Paper Topic Classification (TF-IDF)

This notebook demonstrates preprocessing text, TF-IDF feature extraction, training a classifier, and showing metrics and confusion matrix. It uses a synthetic small dataset for reproducibility.

```
# Imports
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Synthetic dataset: short abstracts and labels (cs, physics, math)
docs = [
    "We present a new sorting algorithm and analyze its complexity.",
    "Experimental results show improved solar cell efficiency using organic materials.",
    "We prove a new theorem in algebraic topology about homology groups.",
    "Neural networks applied to image classification achieve state of the art.",
    "Quantum entanglement measurements and photon experiments are discussed.",
    "We introduce a novel optimization algorithm for convex functions.",
    "Synthesis of nanomaterials for sensors with high sensitivity.",
    "Topology and manifold theory with applications to geometry.",
    "Convolutional models for natural language processing tasks."
]
labels = ["cs","physics","math","cs","physics","cs","physics","math","cs"]

# TF-IDF and classification
vect = TfidfVectorizer()
X = vect.fit_transform(docs)
X_train, X_test, y_train, y_test = train_test_split(X, labels, test_size=0.33, random_state=42, stratify=labels)
clf = LogisticRegression(max_iter=1000)
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
print('Classification Report:\n', classification_report(y_test, y_pred))
cm = confusion_matrix(y_test, y_pred, labels=['cs','physics','math'])
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['cs','physics','math'])
fig = disp.plot().figure_; fig.suptitle('Paper Topic Classification Confusion Matrix'); fig.savefig('task5_confusion.png')
```
