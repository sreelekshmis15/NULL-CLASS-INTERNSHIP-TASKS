```python
# Dynamic Knowledge Expansion Chatbot 

import faiss
import time
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# ------------------------------------------
# Step 1: Load Embedding Model
# ------------------------------------------
model = SentenceTransformer('all-MiniLM-L6-v2')

# ------------------------------------------
# Step 2: Setup vector store (FAISS)
# ------------------------------------------
dimension = 384  # Embedding size of the model
index = faiss.IndexFlatL2(dimension)
knowledge_texts = []  # To keep original text

# ------------------------------------------
# Function 1: Add new knowledge
# ------------------------------------------
def add_to_knowledge_base(new_text_list):
    global knowledge_texts
    embeddings = model.encode(new_text_list)
    index.add(embeddings.astype('float32'))
    knowledge_texts.extend(new_text_list)

# ------------------------------------------
# Step 3: Initial Knowledge (Training Project Dataset Example)
# Replace this with your own project data
# ------------------------------------------
initial_data = [
    "Satellite images are classified into four categories: cloudy, desert, green area, and water.",
    "Convolutional Neural Networks are used to extract spatial features from images.",
    "Data augmentation improves generalization by creating variations of images."
]

add_to_knowledge_base(initial_data)

# ------------------------------------------
# Function 2: Search & Answer
# ------------------------------------------
def chatbot(query):
    query_embedding = model.encode([query])
    distances, indices = index.search(query_embedding.astype('float32'), k=1)
    best_match = knowledge_texts[indices[0][0]]
    return f"Answer: {best_match}"

# ------------------------------------------
# Function 3: Dynamic Data Fetching (Mock Example)
# Replace this with API / GitHub / ArXiv fetch code
# ------------------------------------------
def fetch_new_information():
    # For example demonstration: new research update came
    return [
        "Transfer learning helps improve classification accuracy when training data is limited.",
        "CNN filters detect edges, textures, and patterns in satellite imagery."
    ]

# ------------------------------------------
# Function 4: Periodic Update Mechanism
# ------------------------------------------
def update_knowledge_periodically(interval_seconds=10):
    while True:
        print("\nFetching new knowledge...")
        new_data = fetch_new_information()
        add_to_knowledge_base(new_data)
        print("Knowledge Base Updated!")
        time.sleep(interval_seconds)  # Repeat update every X seconds

# ------------------------------------------
# Example Chat
# ------------------------------------------
print(chatbot("How do CNNs help in satellite image classification?"))

# ------------------------------------------
# Run this only if you want auto updates (Ctrl+C to stop)
# update_knowledge_periodically()

```
